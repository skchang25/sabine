```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from astropy import units as u
from astropy import constants as const
```

```python
data = pd.read_csv("solar_system.csv")
print(data)
print(data.shape) # 20 rows and 11 columns
```

```python
data = data.set_index("Attribute").T
#print(data.shape) # The first column becomes the column of planets, generating 10 rows (one for each planet) and all planet attributes are now in columns.
data.index.name = "Planet"
#print(data.shape) # sets the name of the rows column to be "Planets", the former title of all attributes of the planets are still connected to the attributes title, therefore no column names exist with Planet.
data.reset_index(inplace = True)
#print(data.shape) # resets the first row to be an index row of Attribute and assigns Planet to the column of planet names, Mass to the colum of masses and so forth.
data.columns.name = None
#print(data.shape) # Removes the column name for the index row and no column for attribute exists, leaving each planet as a row, with all attributes connected to their planets
# The original data frame had 20 rows, including 20 attributes and the planet names/ attributes column name and a column for each planet includign the index (11 columns). The 
#     transposed matrix no longer has a column of attributes, the attribtues become the column heads therefore transposed matrix has 10 rows, one for each planet and the Planet title, and 20 columsn for
#     the 20 attributes columns and the index column.
print(data.columns)
# Columns with units: Mass (10^24kg),Diameter (km),Density (kg/m^3),Gravity (m/s^2),Escape Velocity (km/s),Rotation Period (hours),Length of Day (hours),Distance from Sun (10^6 km),Perihelion (10^6 km),Aphelion (10^6 km),Orbital Period (days),Orbital Velocity (km/s),Orbital Inclination (deg),Obliquity to Orbit (deg),Mean Temperature (C),Surface Pressure (bars)
# Columns without units: Orbital Eccentricity, Number of Moons, Ring System?, Global Magnetic Field?
print(data)
```

```python
#for col in data:
   #print(f" {col}: {data[col].dtypes}")
for col in data:
    print(f" {col}: {data[col].apply(type)}")
# All the values within the columns are stored as strings, meaning we cannot arithmatically manipulate or plot the values as they would only respond to string manipulations.
```

```python
data["Mass (10^24kg)"] = pd.to_numeric(data["Mass (10^24kg)"], errors = "coerce")
data["Diameter (km)"] = pd.to_numeric(data["Diameter (km)"], errors = "coerce")
data["Density (kg/m^3)"] = pd.to_numeric(data["Density (kg/m^3)"], errors = "coerce")
data["Gravity (m/s^2)"] = pd.to_numeric(data["Gravity (m/s^2)"], errors = "coerce")
data["Escape Velocity (km/s)"] = pd.to_numeric(data["Escape Velocity (km/s)"], errors = "coerce")
data["Rotation Period (hours)"] = pd.to_numeric(data["Rotation Period (hours)"], errors = "coerce")
data["Length of Day (hours)"] = pd.to_numeric(data["Length of Day (hours)"], errors = "coerce")
data["Distance from Sun (10^6 km)"] = pd.to_numeric(data["Distance from Sun (10^6 km)"], errors = "coerce")
data["Perihelion (10^6 km)"] = pd.to_numeric(data["Perihelion (10^6 km)"], errors = "coerce")
data["Aphelion (10^6 km)"] = pd.to_numeric(data["Aphelion (10^6 km)"], errors = "coerce")
data["Orbital Period (days)"] = pd.to_numeric(data["Orbital Period (days)"], errors = "coerce")
data["Orbital Velocity (km/s)"] = pd.to_numeric(data["Orbital Velocity (km/s)"], errors = "coerce")
data["Orbital Inclination (deg)"] = pd.to_numeric(data["Orbital Inclination (deg)"], errors = "coerce")
data["Orbital Eccentricity"] = pd.to_numeric(data["Orbital Eccentricity"], errors = "coerce")
data["Obliquity to Orbit (deg)"] = pd.to_numeric(data["Obliquity to Orbit (deg)"], errors = "coerce")
data["Mean Temperature (C)"] = pd.to_numeric(data["Mean Temperature (C)"], errors = "coerce")
data["Surface Pressure (bars)"] = pd.to_numeric(data["Surface Pressure (bars)"], errors = "coerce")
data["Number of Moons"] = pd.to_numeric(data["Number of Moons"], errors = "coerce")

data.dtypes

# The three data types are: float64, int64 and object. float64 numerical values with decimal points in 64 bit memory, int64 holds whole number values also in 64 bit memory and object for the categories that were not reassigned. 
#     The object datatype indicates the information is not stored directly in the variable, such as primitive data is, rather it stores an address refrencing the data in a different space. As the three colums were not converted intor primative data types,
#     (Global Magnetic Feild?, Ring System? and Planet) they do not directly contain string data, rather references to data elsewhere.
```

```python
def attach_units(col_name):
    start = col_name.find('(')+1
    end = col_name.find(')')
    sub = col_name[start:end]
    if "kg" in sub:
        if "m" in sub:
            unit = u.kg / u.m **3
            new_col = "Density (kg/m^3)"
            new_values = []
            for val in data[col_name]:
                new_values.append(val*unit)
            data[col_name] = new_values
            data.rename(columns={col_name: new_col}, inplace=True)
        sub = sub[3:-2]
        pre = int(sub)
        unit = 10**pre*u.kg
        new_col = "Mass (kg)"
        new_values = []
        for val in data[col_name]:
            new_values.append(val*unit)
        data[col_name] = new_values
        data.rename(columns={col_name: new_col}, inplace=True)
    if "m/s^2" in sub:
        unit = u.m / u.s ** 2
        new_col = "Gravity (m/s^2)"
        new_values = []
        for val in data[col_name]:
            new_values.append(val*unit)
        data[col_name] = new_values
        data.rename(columns={col_name: new_col}, inplace=True)
    if "km" in sub:
        if "s" in sub:
            unit = u.kg / u.s
            new_col = col_name
            new_values = []
            for val in data[col_name]:
                new_values.append(val*unit)
            data[col_name] = new_values
            data.rename(columns={col_name: new_col}, inplace=True)
        unit = 10**6*u.km
        new_col = col_name[:-9] + "(km)"
        new_values = []
        for val in data[col_name]:
            new_values.append(val*unit)
        data[col_name] = new_values
        data.rename(columns={col_name: new_col}, inplace=True)
    if "hours" in sub:
        unit = u.hour
        new_col = col_name
        new_values = []
        for val in data[col_name]:
            new_values.append(val*unit)
        data[col_name] = new_values
        data.rename(columns={col_name: new_col}, inplace=True)
        print(data)
    if "deg" in sub:
        unit = u.deg
        new_col = col_name
        new_values = []
        for val in data[col_name]:
            new_values.append(val*unit)
        data[col_name] = new_values
        data.rename(columns={col_name: new_col}, inplace=True)
    if "bars" in sub:
        unit = u.bar
        new_col = col_name
        new_values = []
        for val in data[col_name]:
            new_values.append(val*unit)
        data[col_name] = new_values
        data.rename(columns={col_name: new_col}, inplace=True)
    if "C" in sub:
        unit = u.deg_C
        new_col = col_name
        new_values = []
        for val in data[col_name]:
            new_values.append(val*unit)
        data[col_name] = new_values
        data.rename(columns={col_name: new_col}, inplace=True)
    if "days" in sub:
        unit = u.d
        new_col = col_name
        new_values = []
        for val in data[col_name]:
            new_values.append(val*unit)
        data[col_name] = new_values
        data.rename(columns={col_name: new_col}, inplace=True)
attach_units("Surface Pressure (bars)")
print(data["Surface Pressure (bars)"])
```

```python
semi_major_axis = (data["Aphelion (10^6 km)"] + data["Perihelion (10^6 km)"])/2
data.insert(11,"Semi_major_axis (10^6 km)",semi_major_axis*(u.km*10**6))
attach_units("Semi_major_axis (10^6 km)")
print(data["Semi_major_axis (km)"])
```

```python
attach_units("Orbital Period (days)")
new_col = "Orbital Period (years)"
new_values = []
for val in data["Orbital Period (days)"]:
    new_values.append(val.to(u.yr))
data["Orbital Period (days)"] = new_values
data.rename(columns={"Orbital Period (days)": new_col}, inplace=True)
print(data.columns)
print(data["Orbital Period (years)"])
print(f"Pluto's Orbital Period (Years): {data.iloc[9,12]:.4f}")
```

```python
# AU stands for astronimcal unit, a unit of length in 10^11 meters which measures astronomical distances in the average distance between the sun and the earth.
print(const.au)
print(f"Number of kilometers in one AU: {(const.au).to(u.km)}")
```

```python
def to_AU(col_name):
    start = col_name.find('(')
    sub = col_name[:start]
    new_col = sub + "(AU)"
    new_values = []
    for val in data[col_name]:
        new_values.append(val * u.au)
    data[col_name] = new_values
    data.rename(columns={col_name: new_col}, inplace=True)

to_AU("Diameter (km)")
to_AU("Distance from Sun (10^6 km)")
to_AU("Perihelion (10^6 km)")
to_AU("Aphelion (10^6 km)")
to_AU("Semi_major_axis (km)")
```
