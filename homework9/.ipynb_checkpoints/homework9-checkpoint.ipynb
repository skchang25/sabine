```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy.optimize import curve_fit
```

```python
data = pd.read_csv("noisy_curvefit_data.csv")
print(f"Shape: {data.shape}") # 100 rows and 6 columns 
print(f"Head: {data.head()}")
print(f"Tail: {data.tail()}") # columns names: x , y_noise_1, y_noise_2, y_noise_3, y_noise_4, y_noise_5
```

```python
plt.scatter(data["x"],data["y_noise_1"], label = "noisy data")
plt.xlabel("x")
plt.ylabel("y_noise_1")
plt.title("x versus y_noise_1")
plt.legend()
plt.show()
```

```python

plt.scatter(data["x"],data["y_noise_1"], label = "noisy data")
x = data["x"]
y = data["y_noise_1"]
def model(A,k,x):
    return A * np.exp(-k * x)
p0 = [1,1]
popt,pcov = curve_fit(model,x,y, p0)
Afit, kfit = popt
print(f"Fit : y {Afit:.2f} *e^{-kfit:.2f} *x")
plt.plot(x,(Afit * np.exp (-kfit *x)), color = "red", label = "Best fit line")



p0 = [5,0.8]
popt, pcov = curve_fit(model,x,y,p0)
A_fit,k_fit = popt
plt.plot(x,(A_fit * np.exp (-k_fit *x)), color = "black", linestyle = "--", label = "True Model")
print(f"True model fit: y = {A_fit:.2f} *e^{-k_fit:.2f}*x")


residuals = y-model(Afit,kfit,data["x"])
chi2 = np.sum(residuals**2)
dof = len(x)-len(popt)
chi2_reduced = chi2 / dof

noisy_parameters = pd.DataFrame({'Noise Level':[],'A fit':[],'k fit':[],'Reduced Chi-Squared':[]})
noisy_parameters.loc[len(noisy_parameters)] =["y_noise_1",Afit,kfit,chi2_reduced]


plt.xlabel("x")
plt.ylabel("y_noise_1")
plt.title(f"1: x versus y_noise_1, Reduced chi squared: {chi2_reduced:.2f}")
plt.legend()
plt.show()
```

```python
def fit_model(x,y):
    def model(A,k,x):
        return A * np.exp(-k * x)
    p0 = [1,1]
    popt,pcov = curve_fit(model,x,y, p0)
    Afit, kfit = popt
    
    #print(f"Fit : y {Afit:.2f} *e^{-kfit:.2f} *x")
    p0 = [5,0.8]
    popt, pcov = curve_fit(model,x,y,p0)
    Afit,kfit = popt
    
    #print(f"True model fit: y = {Afit:.2f} *e^{-kfit:.2f}*x")
    residuals = y-model(Afit,kfit,data["x"])
    chi2 = np.sum(residuals**2)
    dof = len(x)-len(popt)
    chi2_reduced = chi2 / dof
    fit_values = pd.DataFrame([Afit,kfit, chi2_reduced])
    return fit_values

fig,axes= plt.subplots(2,2) 
axes=axes.flatten()

cols = ["y_noise_2","y_noise_3","y_noise_4","y_noise_5"]
for idx, (col, ax) in enumerate(zip(cols, axes)):
    ax.scatter(data["x"],data[col], color = "grey", label = "noise data")
    fit_values = fit_model(data["x"],data[col])
    ax.plot(x,(fit_values[0][0] * np.exp (-fit_values[0][1] *x)), color = "red", label = "Best fit line")
    ax.plot(x,(Afit * np.exp (-kfit *x)), color = "black", linestyle = "--", label = "True Model")
    noisy_parameters.loc[len(noisy_parameters)] =[cols[idx],fit_values[0][0],fit_values[0][1],fit_values[0][2]]
    ax.set_title(f"{cols[idx]} : {fit_values[0][2]:.2f}")
    ax.set_xlabel("x")
    ax.set_ylabel(cols[idx])
    ax.legend()

# Increased noise levels led to more deviation from the best fit line (red), the True model (black) did tend to flatten with more noise,
#     and repersents a center of the data moreso as it increasing deviates.
# The reduced chi squared increases; with more noise in the data there is more deviation from the expected result.
```

```python
noisy_parameters
noisy_parameters.to_csv("noisy_parameters.csv", index=False)
```
